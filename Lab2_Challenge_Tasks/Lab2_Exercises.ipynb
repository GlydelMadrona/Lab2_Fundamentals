```python
# CELL 1 — STUDENT CONFIGURATION

ENGINEER = "MADRONA"
STUDENT_ID = "TUPM-26-1133"
SEED_NUM = int(STUDENT_ID[-1])

print("Engineer:", ENGINEER)
print("Student ID:", STUDENT_ID)
print("SEED_NUM:", SEED_NUM)
```

```python
# CELL 2 — GENERATE SENSOR DATA

sensor_data = [
    len(ENGINEER) * SEED_NUM + 23,
    len(ENGINEER) * SEED_NUM + 72,
    SEED_NUM * 10 + 30,
    len(ENGINEER) * 8 + 56,
    SEED_NUM * 10 + 42
]

print("Generated Sensor Data:", sensor_data)
```

```python
# CELL 3 — RECEIVE SENSOR READINGS

received_readings = []

print("Enter 3 sensor readings:")

for number in range(3):
    entry = input(f"Reading {number + 1}: ")
    received_readings.append(entry)

print("Received Readings:", received_readings)
```

```python
# CELL 4 — VALIDATE SENSOR READINGS

def validate_sensor(value):
    try:
        number = float(value)

        if number < 0:
            return False, value

        return True, number

    except ValueError:
        return False, value


validation_results = []

for reading in received_readings:
    validation_results.append(validate_sensor(reading))

print("Validation Results:", validation_results)
```

```python
# CELL 5 — CLASSIFY VALID READINGS

def classify_sensor(value):
    if value <= 100:
        return "NOMINAL"
    else:
        return "OVER LIMIT"


classification_results = []

for valid, value in validation_results:
    if valid:
        classification_results.append(classify_sensor(value))

print("Classification Results:", classification_results)
```

```python
# CELL 6 — CHALLENGE 1 SUMMARY

valid_values = [
    value for valid, value in validation_results
    if valid
]

invalid_values = [
    value for valid, value in validation_results
    if not valid
]

print("Execution Log:")
print("Received sensor readings from operator input.")
print(
    "Valid readings:",
    valid_values
)
print(
    "Invalid readings:",
    invalid_values
)
print(
    "Classification:",
    classification_results
)

print("\nFinal Output:")
print("Sensor Data:", sensor_data)
print("Valid Count:", len(valid_values))
print("Invalid Count:", len(invalid_values))
print("Classification:", classification_results)

if classification_results and all(
    result == "NOMINAL" for result in classification_results
):
    print("Safety Status: SAFE")
else:
    print("Safety Status: REVIEW REQUIRED")
```

```python
# CELL 7 — GENERATE SIGNAL

signal = ENGINEER.upper() + str(SEED_NUM)

print("Generated Signal:", signal)
```

```python
# CELL 8 — NORMALIZE SIGNAL

normalized_signal = signal.strip().upper()

print("Processed Signal:", normalized_signal)
```

```python
# CELL 9 — CHARACTER ANALYSIS

letters = 0
numbers = 0
spaces = 0
other_characters = 0

for character in normalized_signal:

    if character.isalpha():
        letters += 1

    elif character.isdigit():
        numbers += 1

    elif character.isspace():
        spaces += 1

    else:
        other_characters += 1


print("Character Analysis:")
print("Letters:", letters)
print("Numbers:", numbers)
print("Spaces:", spaces)
print("Other Characters:", other_characters)
```

```python
# CELL 10 — SIGNAL CLASSIFICATION

def determine_signal_type(
    letter_count,
    number_count,
    space_count,
    other_count
):

    if (
        letter_count > 0
        and number_count > 0
        and space_count == 0
        and other_count == 0
    ):
        return "ALPHANUMERIC SIGNAL"

    elif letter_count > 0 and number_count == 0:
        return "LETTER SIGNAL"

    elif number_count > 0 and letter_count == 0:
        return "NUMERIC SIGNAL"

    else:
        return "MIXED SIGNAL"


signal_classification = determine_signal_type(
    letters,
    numbers,
    spaces,
    other_characters
)

print("Signal Classification:", signal_classification)
```

```python
# CELL 11 — CHALLENGE 2 FINAL REPORT

print("Execution Log:")
print(
    f"Program generated the signal {signal} "
    f"using ENGINEER and SEED_NUM."
)
print(
    f"Signal was normalized to {normalized_signal}."
)
print(
    f"Character analysis detected {letters} letters, "
    f"{numbers} numbers, {spaces} spaces, "
    f"and {other_characters} other characters."
)

print("\nFinal Output:")
print("SIGNAL:", signal)
print("PROCESSED SIGNAL:", normalized_signal)
print("LETTERS:", letters)
print("NUMBERS:", numbers)
print("SPACES:", spaces)
print("OTHER CHARACTERS:", other_characters)
print("CLASSIFICATION:", signal_classification)
```

```python
# CELL 12 — AUTHENTICATION CONFIGURATION

access_key = ENGINEER.upper() + str(SEED_NUM)

attempt_limit = 10
attempts_made = 0
access_granted = False

print("Generated Password:", access_key)
print("Attempt Limit:", attempt_limit)
```

```python
# CELL 13 — PASSWORD VERIFICATION

def check_credentials(entered_key, authorized_key):
    return entered_key == authorized_key


print("Authentication system ready.")
```

```python
# CELL 14 — AUTHENTICATION PROCESS

while attempts_made < attempt_limit:

    entered_key = input("Enter password: ")
    attempts_made += 1

    if check_credentials(entered_key, access_key):
        access_granted = True
        print("Access Result: ACCESS GRANTED")
        break
```

```python
# CELL 15 — DETERMINE FINAL SYSTEM STATE

if access_granted:
    system_state = "UNLOCKED"
else:
    system_state = "LOCKED"

print("Access Granted:", access_granted)
print("Final System State:", system_state)
```

```python
# CELL 16 — FINAL OUTPUT AND EXECUTION LOG

print("\nExecution Log:")

if access_granted:
    print(
        f"Access was granted after {attempts_made} attempt(s)."
    )
else:
    print(
        f"Access was denied after {attempts_made} attempt(s)."
    )

print("\nFinal Output:")
print("Generated Password:", access_key)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)

if access_granted:
    print("Access Result: ACCESS GRANTED")
else:
    print("Access Result: ACCESS DENIED")

print("Final System State:", system_state)
```
